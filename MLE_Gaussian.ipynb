```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# -----------------------------
# True parameters
# -----------------------------
np.random.seed(0)
mu_true = 5
sigma = 2
n = 200

# -----------------------------
# Simulate many experiments
# -----------------------------
num_trials = 5000
mle_estimates = []

for _ in range(num_trials):
    x = np.random.normal(mu_true, sigma, n)
    mle_estimates.append(np.mean(x))

mle_estimates = np.array(mle_estimates)

# -----------------------------
# Theoretical distribution
# -----------------------------
mu_vals = np.linspace(3, 7, 500)
theoretical_pdf = norm.pdf(mu_vals, mu_true, sigma/np.sqrt(n))

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(8,5))

# Histogram of empirical MLE distribution
plt.hist(mle_estimates, bins=50, density=True, alpha=0.6, label="Empirical")

# Theoretical curve
plt.plot(mu_vals, theoretical_pdf, linewidth=2, label="Theoretical N(mu, sigma^2/n)")

plt.axvline(mu_true, linestyle='--', label="True mean")

plt.title("Distribution of MLE of Mean")
plt.xlabel("Estimated Mean")
plt.ylabel("Density")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

np.random.seed(0)

# True parameters
mu_true = 5
sigma = 2

# Different sample sizes
n_values = [5, 20, 100]

num_trials = 5000
mu_vals = np.linspace(3, 7, 500)

plt.figure(figsize=(8,5))

for n in n_values:
    mle_estimates = []

    for _ in range(num_trials):
        x = np.random.normal(mu_true, sigma, n)
        mle_estimates.append(np.mean(x))

    mle_estimates = np.array(mle_estimates)

    # Plot histogram
    plt.hist(mle_estimates, bins=50, density=True, alpha=0.3, label=f"Empirical n={n}")

    # Theoretical curve
    theoretical_pdf = norm.pdf(mu_vals, mu_true, sigma/np.sqrt(n))
    plt.plot(mu_vals, theoretical_pdf, linewidth=2, label=f"Theory n={n}")

plt.axvline(mu_true, linestyle='--', label="True mean")

plt.title("MLE Distribution for Different Sample Sizes")
plt.xlabel("Estimated Mean")
plt.ylabel("Density")
plt.legend()
plt.show()
```
