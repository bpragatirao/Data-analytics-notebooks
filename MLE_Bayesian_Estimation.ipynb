```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
np.random.seed(727)

mu = 2
sigma = 1

# Generate data
n = 20
x = np.random.normal(mu, sigma, n)
```

```python
# MLE
mu_est_MLE = np.mean(x)

# Prior
mu_0 = 0          # prior mean
sigma_0 = .1      # prior variance

norm = n*(sigma_0**2) + sigma**2
mu_est_Bayes = (n*(sigma_0**2)/norm) * mu_est_MLE + ((sigma**2)/norm) * mu_0  


print(mu_est_MLE)
print(mu_est_Bayes)
```
